````python
# Libraries
import os
import re
import json
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from bs4 import BeautifulSoup

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Get current working directory
print(os.getcwd())
````

### Produktlisten-Seiten scrapen → `df_raw`

Der folgende Code liest die drei gespeicherten Produktlisten-Seiten mit BeautifulSoup ein (analog zur Übung `skipass_data.ipynb`). Jede Produktkarte (`<div class="product-card">`) wird zu einer Zeile im Data Frame `df_raw`. Die Werte werden **als Text** übernommen (Spalten mit Endung `_raw`) und noch nicht bereinigt.

````python
# Initialize list for the extracted rows
rows = []

# Loop over the three listing pages
for page in [1, 2, 3]:

    # Read html from file
    with open(f'html/shop_page_{page}.html', 'r', encoding='utf-8') as file:
        html_content = file.read()

    # Parse the HTML content
    soup = BeautifulSoup(html_content, 'html.parser')

    # Locate the product cards and extract values
    for card in soup.select('div.product-card'):
        row_data = [card['data-product-id'],
                    card.select('span.brand')[0].get_text().replace('\xa0', '').strip(),
                    card.select('span.model')[0].get_text().replace('\xa0', '').strip(),
                    card.select('li.storage')[0].get_text().replace('\xa0', '').strip(),
                    card.select('li.ram')[0].get_text().replace('\xa0', '').strip(),
                    card.select('li.color')[0].get_text().replace('\xa0', '').strip(),
                    card.select('li.screen')[0].get_text().replace('\xa0', '').strip(),
                    card.select('li.weight')[0].get_text().replace('\xa0', '').strip(),
                    card.select('span.price')[0].get_text().replace('\xa0', '').strip(),
                    card.select('span.availability')[0].get_text().replace('\xa0', '').strip()]
        rows.append(row_data)

# Create a DataFrame
df_raw = pd.DataFrame(rows, columns=['product_id', 'brand', 'model', 'storage_raw', 'ram_raw', 'color',
                                     'screen_raw', 'weight_raw', 'price_raw', 'availability'])

# Show dimensions and first rows
print('Number of rows and columns:', df_raw.shape)
df_raw.head()
````

**Ziel:** Kennzahlen für das Shop-Sortiment aus den gescrapten Rohdaten `df_raw`: Anzahl Geräte, Anzahl Geräte ohne Preisangabe und durchschnittliche Speicherkapazität in GB.

```python
# Select columns for the analysis
df_B1 = df_raw[['product_id', 'brand', 'model', 'storage_raw', 'price_raw']]

# Extract numerical values from price_raw
price = []
for i in df_B1['price_raw']:
    d1 = re.findall(r"\d+", i)
    try:
        d2 = int(''.join(d1))
    except:
        d2 = None
    price.append(d2)

# Extract numerical values from storage_raw (in GB)
```

Tatsächliches Ergebnis:

```
Anzahl Geräte:               152
Anzahl ohne Preisangabe:     8
Ø Speicherkapazität (GB):    234.6
```

Der Shop gibt auf jeder Seite an, dass er **146 Artikel** führt. Beispiele für Werte in `storage_raw`: `256GB`, `512GB`, `1TB`, `2TB`. Beispiele für Werte in `price_raw`: `CHF 899.–`, `CHF 2'199.–`, `Preis auf Anfrage`.

**Aufgabe:** Schreiben Sie den korrigierten Code in die Zelle «Korrigierter Code» (die Kommentare des ursprünglichen Codes sind als Gerüst vorgegeben), sodass die Kennzahlen stimmen. Begründen Sie jede Änderung gegenüber dem ursprünglichen Code.

**Begründung der Änderungen:**

1. **Duplikate entfernen:** Es gibt Duplikate (152 statt 146 Geräte), weil sich die Seiten überlappen (die letzten 3 Karten einer Seite erscheinen auf der nächsten Seite erneut). `drop_duplicates()` entfernt sie, `reset_index(drop=True)` nummeriert den Index neu. Da eines der doppelten Geräte «Preis auf Anfrage» hat, ist auch die Anzahl ohne Preisangabe zu hoch (8 statt 7).
2. **Einheiten vereinheitlichen:** Bei `1TB` bzw. `2TB` liefert `d1[0]` den Wert `1` bzw. `2` — als wären es GB. Werte in TB müssen mit 1024 multipliziert werden (gemäss Spaltenbeschreibung: 1 TB = 1024 GB). Sonst wird der Durchschnitt massiv unterschätzt.

Ergebnis: **146** Geräte, **7** ohne Preisangabe, Ø Speicherkapazität **783.8 GB**.

**Korrigierter Code:**

````python
# Select columns for the analysis, remove duplicated values and reset the index
df_B1 = df_raw[['product_id', 'brand', 'model', 'storage_raw', 'price_raw']].drop_duplicates().reset_index(drop=True)

# Extract numerical values from price_raw
price = []
for i in df_B1['price_raw']:
    d1 = re.findall(r"\d+", i)
    try:
        d2 = int(''.join(d1))
    except:
        d2 = None
    price.append(d2)

# Extract numerical values from storage_raw (in GB, note the conversion of TB)
storage = []
for i in df_B1['storage_raw']:
    d1 = re.findall(r"\d+", i)
    if 'TB' in i:
        storage.append(int(d1[0]) * 1024)
    else:
        storage.append(int(d1[0]))

# Save as new variables in the pandas data frame
df_B1['price_chf'] = pd.Series(price, dtype="Int64")
df_B1['storage_gb'] = storage

# Show key figures
print('Anzahl Geräte:              ', df_B1.shape[0])
print('Anzahl ohne Preisangabe:    ', df_B1['price_chf'].isnull().sum())
print('Ø Speicherkapazität (GB):   ', round(df_B1['storage_gb'].mean(), 1))
````

«Preisverlauf» ist die **zweite** Tabelle der Seite (die erste ist «Weitere Varianten»), daher `find_all('table')[1]` — alternativ eindeutig über die Klasse: `soup.find('table', class_='price-history')`. Die Kopfzeile enthält nur `<th>`-Zellen und ergibt eine leere Zeile, die `dropna()` entfernt; die Ziffern werden mit `re.findall` extrahiert und mit `''.join` zusammengefügt (wegen des Apostrophs).

**Python-Code:**

````python
# Read html from file
with open('html/product_detail.html', 'r', encoding='utf-8') as file:
    html_content = file.read()

# Parse the HTML content
soup = BeautifulSoup(html_content, 'html.parser')

# Locate the table «Preisverlauf» (second table on the page)
table = soup.find_all('table')[1]

# Extract table rows
rows = []
for row in table.find_all('tr'):
    rows.append([cell.get_text().strip() for cell in row.find_all('td')])

# Create a DataFrame
df_C1 = pd.DataFrame(rows, columns=['datum', 'preis'])

# Remove the empty header row and convert the price to an integer
df_C1 = df_C1.dropna().reset_index(drop=True)
df_C1['preis_chf'] = [int(''.join(re.findall(r"\d+", i))) for i in df_C1['preis']]

# Show the result
print(df_C1.dtypes, '\n')
df_C1
````

### C2 (4 Punkte)

Teilen Sie die Geräte in `df_clean` anhand von `screen_inch` mit `pd.cut()` in **drei Grössenklassen** ein (neue Spalte `klasse`):

- `kompakt`: unter 14 Zoll
- `standard`: 14 Zoll bis unter 16 Zoll
- `gross`: ab 16 Zoll

Erstellen Sie anschliessend mit `pd.crosstab()` eine Kontingenztabelle mit dem **Anteil in Prozent** der Geräte pro Marke (`brand`, Zeilen) und Grössenklasse (`klasse`, Spalten), sodass sich jede Zeile auf 100% summiert (gerundet auf 1 Dezimalstelle). Begründen Sie, wie Sie die Klassengrenzen festlegen (insbesondere für Geräte mit genau 14.0 bzw. 16.0 Zoll) und warum Anteile pro Marke aussagekräftiger sind als absolute Anzahlen.

**Begründung der Vorgehensweise:**

`pd.cut()` verwendet **feste**, inhaltlich vorgegebene Grenzen (anders als `pd.qcut()`, das Quantile verwendet — die Klassen sind daher unterschiedlich gross: 16 / 104 / 26 Geräte). Mit `right=False` sind die Intervalle links geschlossen, `[14, 16)`: Ein Gerät mit genau 14.0 Zoll ist `standard`, eines mit 16.0 Zoll `gross`. Mit `normalize='index'` werden die Anteile pro Zeile berechnet; das ist aussagekräftiger, weil die Marken sehr unterschiedlich viele Geräte führen (Apple 38, Framework 2).

**Python-Code:**

````python
# Select columns for the analysis
df_C2 = df_clean[['brand', 'screen_inch']].copy()

# Discretization of the variable 'screen_inch' into 3 bins with fixed limits (note right=False)
df_C2['klasse'] = pd.cut(df_C2['screen_inch'],
                         bins=[0, 14, 16, 20],
                         right=False,
                         labels=['kompakt', 'standard', 'gross'])

# Check the class sizes
print(df_C2['klasse'].value_counts(), '\n')

# Contingency table with row percentages
(pd.crosstab(df_C2['brand'], df_C2['klasse'], normalize='index') * 100).round(1)
````

Die Datei `ratings.json` ist **verschachtelt** aufgebaut (`meta` mit Angaben zur API-Antwort, `data` mit den Bewertungen; Anzahl und Durchschnitt liegen im Unterobjekt `rating`). Lesen Sie die Bewertungen mit `json.load()` und `pd.json_normalize()` in einen flachen Data Frame ein und verbinden Sie ihn mit `.merge()` über `product_id` so mit `df_clean`, dass **alle Geräte des Shops** erhalten bleiben. Geben Sie aus, wie viele Geräte keine Bewertung haben.

Berechnen Sie die **Spearman-Rangkorrelation** zwischen `weight_kg` und `rating.avg` (gerundet auf 2 Dezimalstellen) und erstellen Sie ein Streudiagramm (Scatter Plot) mit beschrifteten Achsen.

Begründen Sie, warum Sie `pd.json_normalize()` und den gewählten Join-Typ verwenden, und interpretieren Sie das Ergebnis in 1–2 Sätzen.

**Begründung der Vorgehensweise:**

`pd.json_normalize()` macht aus den verschachtelten Einträgen unter `data` flache Spalten (`rating.count`, `rating.avg`); `meta` enthält keine Beobachtungen und wird nicht übernommen. Mit `how='left'` bleiben alle **146** Geräte erhalten; **18** Geräte ohne Bewertung erhalten `NaN`, die 6 Bewertungen nicht mehr geführter Geräte fallen weg. Die Korrelation beruht auf **120** Geräten (ohne Bewertung bzw. ohne Gewicht, d.h. Acer, werden ausgeschlossen).

**Interpretation:** Mit −0.49 besteht ein mittlerer negativer Zusammenhang: Leichtere Geräte werden tendenziell besser bewertet. Das ist kein Beleg für Ursache und Wirkung (z.B. sind leichte Geräte oft auch hochwertiger). Spearman ist hier sinnvoll, weil er robust gegenüber dem Erfassungsfehler von 9.90 kg ist (Pearson: −0.09).

**Python-Code:**

````python
# Read the nested json file
with open('ratings.json', 'r', encoding='utf-8') as file:
    raw = json.load(file)

# Flatten the entries under 'data' (rating.count, rating.avg)
df_ratings = pd.json_normalize(raw['data'])
print(df_ratings.head(3), '\n')

# Join the ratings to the shop data (keep all devices of the shop)
df_C3 = df_clean.merge(df_ratings, on='product_id', how='left')
print('Number of rows:', df_C3.shape[0])
print('Devices without rating:', df_C3['rating.avg'].isnull().sum(), '\n')

# Spearman rank correlation
print(df_C3[['weight_kg', 'rating.avg']].corr(method='spearman').round(2))

# Exclude the data entry error of 9.90 kg from B2 (for readability of the plot)
df_plot = df_C3.loc[(df_C3['weight_kg'] != 9.9)]

# Scatterplot of weight and customer rating
plt.figure(figsize=(7,4))
plt.scatter(df_plot['weight_kg'], df_plot['rating.avg'])
plt.title('Gewicht vs. Kundenbewertung', fontsize=12)
plt.xlabel('Gewicht (kg)')
plt.ylabel('Durchschnittliche Bewertung (1–5 Sterne)')
plt.show()
````
